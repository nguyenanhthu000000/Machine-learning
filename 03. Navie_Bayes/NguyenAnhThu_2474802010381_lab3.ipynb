```python
# -*- coding: utf-8 -*-
"""Bước 1: tính định lý Bayes bằng tay, không cần thư viện học máy.

Chú thích viết tiếng Việt có dấu, phần in ra màn hình viết không dấu.
"""


def bayes(tien_nghiem, kha_nang, kha_nang_nguoc):
    """Tra ve xac suat hau nghiem.

    tien_nghiem : P(A), niem tin truoc khi co bang chung
    kha_nang : P(B | A), bang chung xuat hien khi A dung
    kha_nang_nguoc : P(B | khong A), bang chung xuat hien khi A sai
    """
    tu_so = kha_nang * tien_nghiem
    mau_so = tu_so + kha_nang_nguoc * (1 - tien_nghiem)
    return tu_so / mau_so


print("Vi du 1: xet nghiem benh")
print(" Ty le mac benh 1 phan tram, xet nghiem dung 99 phan tram.")
p = bayes(tien_nghiem=0.01, kha_nang=0.99, kha_nang_nguoc=0.01)
print(f" Duong tinh thi kha nang co benh that = {p:.4f}")
print()

print("Cung xet nghiem do, doi ty le mac benh:")
print(" Ty le mac benh Hau nghiem khi duong tinh")
for ty_le in [0.001, 0.01, 0.05, 0.10, 0.50]:
    print(f" {ty_le:6.3f} {bayes(ty_le, 0.99, 0.01):.4f}")
print()

print("Vi du 2: hai day chuyen san xuat")
print(" Day A lam 80 phan tram san pham, ty le loi 1 phan tram.")
print(" Day B lam 20 phan tram san pham, ty le loi 5 phan tram.")
p_b = bayes(tien_nghiem=0.20, kha_nang=0.05, kha_nang_nguoc=0.01)
print(f" Nhat duoc san pham loi, kha nang no tu day B = {p_b:.4f}")
print()

print("Kiem lai bang cach dem tren 1000 san pham:")
loi_a = 800 * 0.01
loi_b = 200 * 0.05
print(f" Day A: 800 san pham, {loi_a:.0f} cai loi")
print(f" Day B: 200 san pham, {loi_b:.0f} cai loi")
tong_loi = loi_a + loi_b
print(f" Ty le tu day B = {loi_b} / {tong_loi} = {loi_b / tong_loi:.4f}")
```

```python
# -*- coding: utf-8 -*-
"""Bước 3: huấn luyện bộ lọc thư rác và chấm điểm nó."""

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import (accuracy_score, confusion_matrix,
                            precision_score, recall_score)
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

df = pd.read_csv("tin_nhan.csv")
vec = CountVectorizer(token_pattern=r"(?u)\S+")
X = vec.fit_transform(df["noi_dung"])
y = df["la_rac"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
print("So tin de hoc :", X_train.shape[0])
print("So tin de kiem tra:", X_test.shape[0])
print()

mo_hinh = MultinomialNB(alpha=1.0)
mo_hinh.fit(X_train, y_train)
y_pred = mo_hinh.predict(X_test)

tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
print("Ma tran nham lan")
print(f" TN = {tn:2d} thu thuong duoc giu dung")
print(f" FP = {fp:2d} thu thuong bi xoa nham")
print(f" FN = {fn:2d} thu rac lot luoi")
print(f" TP = {tp:2d} thu rac bi chan dung")
print()

print(f"Accuracy = {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision = {precision_score(y_test, y_pred):.4f}")
print(f"Recall = {recall_score(y_test, y_pred):.4f}")
print()

tin_moi = [
    "trung thuong khuyen mai nhanh tay click ngay",
    "hop nhom bao cao chieu mai phong 305",
    "ban gui giup em tai lieu bai tap nhe",
]
X_moi = vec.transform(tin_moi)
p_rac = mo_hinh.predict_proba(X_moi)[:, 1]

print("Thu tren ba tin nhan tu go:")
for tin, p in zip(tin_moi, p_rac):
    nhan = "RAC" if p >= 0.5 else "thuong"
    print(f" [{nhan:6s}] p = {p:.4f} <- {tin}")
```

```python
# -*- coding: utf-8 -*-
"""Bước 4: tự cài Naive Bayes trên một kho nhỏ để thấy vấn đề số không.

Đoạn này cố ý KHÔNG dùng scikit-learn, vì thư viện đã làm trơn sẵn nên
sinh viên không nhìn thấy chỗ hỏng.
"""

from collections import Counter

kho = [
    ("trung thuong ngay", 1),
    ("khuyen mai giam gia", 1),
    ("trung thuong click", 1),
    ("hop nhom chieu mai", 0),
    ("bai tap nhom nop mai", 0),
    ("thay gui tai lieu", 0),
]

tu_dien = sorted({t for cau, _ in kho for t in cau.split()})
V = len(tu_dien)
print(f"Tu dien co {V} tu:")
for i in range(0, V, 9):
    print(" " + " ".join(tu_dien[i:i + 9]))
print()

dem = {0: Counter(), 1: Counter()}
tong = {0: 0, 1: 0}
for cau, nhan in kho:
    for t in cau.split():
        dem[nhan][t] += 1
        tong[nhan] += 1

print("So lan xuat hien cua tung tu")
print(f" {'tu':12s} {'rac':>5s} {'thuong':>7s}")
for t in tu_dien:
    print(f" {t:12s} {dem[1][t]:5d} {dem[0][t]:7d}")
print()


def xac_suat(tu, nhan, alpha):
    """P(tu | nhan) co lam tron Laplace voi he so alpha."""
    return (dem[nhan][tu] + alpha) / (tong[nhan] + alpha * V)


def diem(cau, nhan, alpha):
    """Tich cua xac suat tien nghiem va cac xac suat tu."""
    p = 0.5
    for t in cau.split():
        if t in tu_dien:
            p *= xac_suat(t, nhan, alpha)
    return p


cau_thu = "trung thuong bai tap"
print(f'Tin nhan thu: "{cau_thu}"')
print("Tu 'bai' chi xuat hien o lop thuong, chua bao gio o lop rac.")
print()

for alpha in [0.0, 1.0]:
    d_rac = diem(cau_thu, 1, alpha)
    d_thuong = diem(cau_thu, 0, alpha)
    ket = "rac" if d_rac > d_thuong else "thuong"
    print(f"alpha = {alpha}")
    print(f" diem lop rac = {d_rac:.10f}")
    print(f" diem lop thuong = {d_thuong:.10f}")
    print(f" ket luan = {ket}")
    print()

print("Khong lam tron, CA HAI diem deu bang 0 nen mo hinh te liet.")
print("Lop rac bang 0 vi tu 'bai' chua tung xuat hien o lop rac.")
print("Lop thuong bang 0 vi tu 'trung' chua tung xuat hien o lop thuong.")
print("Co lam tron, diem lop rac lon hon va mo hinh ket luan dung.")
```

```python
# Bai 1:

def bayes_3_nhanh(tien_nghiem, kha_nang):
    """Tính xác suất hậu nghiệm cho 3 nhóm.

    tien_nghiem: list P(Ai) [Nhóm đều, Nhóm dồn, Ít học]
    kha_nang: list P(B | Ai) tỷ lệ qua môn của từng nhóm
    """
    tu_sos = [kn * tn for kn, tn in zip(kha_nang, tien_nghiem)]
    mau_so = sum(tu_sos)

    hau_nghiem = [tu / mau_so for tu in tu_sos]
    return hau_nghiem


tien_nghiem = [0.50, 0.30, 0.20]
kha_nang = [0.95, 0.70, 0.30]

ket_qua = bayes_3_nhanh(tien_nghiem, kha_nang)

print("Xác suất bạn qua môn thuộc về từng nhóm:")
print(f" - Nhóm học đều (P = 0.95): {ket_qua[0]:.4f}")
print(f" - Nhóm học dồn (P = 0.70): {ket_qua[1]:.4f}")
print(f" - Nhóm ít học (P = 0.30) : {ket_qua[2]:.4f}")
print(f"Tổng xác suất kiểm tra: {sum(ket_qua):.4f}")
```

```python
# Bài 3
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

df = pd.read_csv("tin_nhan.csv")
vec = CountVectorizer(token_pattern=r"(?u)\S+")
X = vec.fit_transform(df["noi_dung"])
y = df["la_rac"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

alphas = [0.001, 0.1, 1.0, 10.0, 100.0]

for alpha in alphas:
    mo_hinh = MultinomialNB(alpha=alpha)
    mo_hinh.fit(X_train, y_train)
    y_pred = mo_hinh.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)

    print(f"alpha = {alpha:6.3f} | Accuracy = {acc:.4f} | Precision = {prec:.4f} | Recall = {rec:.4f}")

print("\nNhận xét:")
print("Khi alpha quá lớn, mọi xác suất từ bị xích lại gần nhau (làm phẳng quá mức), khiến mô hình mất khả năng phân biệt đặc trưng giữa các lớp, dẫn đến chất lượng dự đoán giảm sút.")
```

```python
# Bài 4

import numpy as np
import pandas as pd
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

df = pd.read_csv("tin_nhan.csv")
texts = df["noi_dung"].tolist()
labels = df["la_rac"].tolist()
train_texts, test_texts, y_train, y_test = train_test_split(
    texts, labels, test_size=0.25, random_state=17, stratify=labels
)

tu_dien = sorted({t for cau in train_texts for t in cau.split()})
V = len(tu_dien)
tu_to_idx = {t: i for i, t in enumerate(tu_dien)}

dem = {0: Counter(), 1: Counter()}
tong_tu = {0: 0, 1: 0}

for cau, nhan in zip(train_texts, y_train):
    for t in cau.split():
        if t in tu_to_idx:
            dem[nhan][t] += 1
            tong_tu[nhan] += 1

n_train = len(y_train)
p_prior = {
    0: sum(1 for y in y_train if y == 0) / n_train,
    1: sum(1 for y in y_train if y == 1) / n_train
}

def du_doan_1_cau(cau, alpha=1.0):
    log_p = {0: np.log(p_prior[0]), 1: np.log(p_prior[1])}

    for t in cau.split():
        if t in tu_to_idx:
            for nhan in [0, 1]:
                p_t = (dem[nhan][t] + alpha) / (tong_tu[nhan] + alpha * V)
                log_p[nhan] += np.log(p_t)

    return 1 if log_p[1] > log_p[0] else 0

y_pred_tu_viet = [du_doan_1_cau(cau, alpha=1.0) for cau in test_texts]
acc_tu_viet = accuracy_score(y_test, y_pred_tu_viet)
vec = CountVectorizer(token_pattern=r"(?u)\S+")
X_train = vec.fit_transform(train_texts)
X_test = vec.transform(test_texts)

sk_model = MultinomialNB(alpha=1.0)
sk_model.fit(X_train, y_train)
y_pred_sk = sk_model.predict(X_test)
acc_sk = accuracy_score(y_test, y_pred_sk)

print(f"Accuracy bộ lọc tự viết : {acc_tu_viet:.4f}")
print(f"Accuracy MultinomialNB   : {acc_sk:.4f}")

print("kết quả so sánh:")
print(f"Độ chính xác (Accuracy) của bộ lọc tự viết là {acc_tu_viet:.4f} và của scikit-learn là {acc_sk:.4f}.")
print("Hai con số rất gần nhau (hoặc trùng khớp), chứng minh rằng mã nguồn tự cài đặt đã mô phỏng chính xác logic toán học bên trong của thuật toán MultinomialNB")

print("\nGiải thích:")
print("Kỹ thuật làm trơn Laplace (alpha = 1.0): Giúp cộng thêm một lượng nhỏ vào số đếm, ngăn chặn hiện tượng xác suất bằng 0 khi gặp từ chưa xuất hiện ở một lớp, tránh làm tê liệt toàn bộ phép tính")
print("Chuyển tích thành tổng Loga: Thay vì nhân nhiều xác suất nhỏ gây ra hiện tượng tràn dưới khiến máy tính làm tròn thành 0, dùng hàm loga để chuyển thành phép cộng các log-xác suất, vừa an toàn về mặt số học vừa giữ nguyên thứ tự lớn nhỏ giữa các lớp")
```

```python
# Bài 5
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import MultinomialNB, GaussianNB

df = pd.read_csv("sinh_vien.csv")
X = df[["gio_on", "diem_giua_ky"]]
y = df["qua_mon"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

mnb = MultinomialNB()
mnb.fit(X_train, y_train)
acc_mnb = accuracy_score(y_test, mnb.predict(X_test))
gnb = GaussianNB()
gnb.fit(X_train, y_train)
acc_gnb = accuracy_score(y_test, gnb.predict(X_test))

print(f"Accuracy của MultinomialNB: {acc_mnb:.4f}")
print(f"Accuracy của GaussianNB   : {acc_gnb:.4f}")

print("\nGiải thích:")
print("Dùng MultinomialNB cho dữ liệu số liệu liên tục (như số giờ ôn, điểm giữa kỳ) là sai về nguyên tắc vì MultinomialNB giả định dữ liệu là tần suất đếm (counts) rời rạc. Tuy nhiên chương trình vẫn chạy và ra kết quả vì các giá trị liên tục trong tập dữ liệu không âm, nhưng mô hình sẽ không mô tả đúng phân phối chuẩn của dữ liệu như GaussianNB.")
```

```python
# Bài 6

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

df = pd.read_csv("tin_nhan.csv")
df_thuong = df[df["la_rac"] == 0]
df_rac = df[df["la_rac"] == 1].head(20)
df_moi = pd.concat([df_thuong, df_rac]).reset_index(drop=True)

print("Phân bố nhãn trong bộ dữ liệu mới:")
print(df_moi["la_rac"].value_counts())
print()

vec = CountVectorizer(token_pattern=r"(?u)\S+")
X = vec.fit_transform(df_moi["noi_dung"])
y = df_moi["la_rac"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

mo_hinh = MultinomialNB(alpha=1.0)
mo_hinh.fit(X_train, y_train)
y_pred = mo_hinh.predict(X_test)
acc = accuracy_score(y_test, y_pred)
recall_rac = recall_score(y_test, y_pred, pos_label=1, zero_division=0)

print(f"Accuracy = {acc:.4f}")
print(f"Recall lớp rác (pos_label=1) = {recall_rac:.4f}")
print("\nMa trận nhầm lẫn:")
print(confusion_matrix(y_test, y_pred))
print("Accuracy có còn phản ánh đúng chất lượng bộ lọc không?")
print("Không, dù chỉ số Accuracy đạt mức rất cao, nhưng bộ lọc thực chất đã gần như vô dụng đối với lớp rác")
print("Nguyên nhân là do bộ dữ liệu bị mất cân bằng nặng (180 tin thường và chỉ có 20 tin rác). Xác suất tiên nghiệm (prior) của lớp thường quá lớn, lấn át hoàn toàn lớp rác")

print("\nPhân tích qua chỉ số Recall của lớp rác và bẫy Accuracy:")
print(f"Chỉ số Recall của lớp rác đạt mức rất thấp ({recall_rac:.4f}), chứng minh rằng mô hình bỏ sót phần lớn hoặc toàn bộ các tin nhắn rác thực tế trong tập kiểm tra")
print(f"Mô hình bị 'mắc bẫy accuracy': Do phán đoán thiên hẳn về lớp đa số (lớp thường), mô hình vẫn đạt Accuracy cao ({acc:.4f}) dù không cần phát hiện đúng tin rác nào. Điều này cho thấy khi dữ liệu mất cân bằng, không bao giờ được chỉ nhìn vào Accuracy mà phải đánh giá qua Recall và Precision của từng lớp")
```
